## Naming convention

Dataset names follow **material-first** convention: `{material}_{descriptor}`.

| Rule | Example |
|------|---------|
| Lowercase, underscores only | `srtio3_lamella` |
| Material first | `gold_nanoparticle`, `graphene_monolayer` |
| No year, resolution, or binning | Put those in metadata JSON |
| Lab suffix only to disambiguate | `srtio3_lamella_ncem` vs `srtio3_lamella_oxford` |

## Technique folders

| Folder | Data type |
|--------|-----------|
| `4dstem` | 4D-STEM diffraction |
| `hrtem` | High-resolution TEM |
| `eels` | Electron energy loss |
| `tomo` | Tomography |
| `diffraction` | Diffraction patterns |
| `image` | Virtual/derived images |
| `complex` | Ptychography |

## Step 1: Prepare your data

```python
import numpy as np

# Replace with your real data
data = np.random.rand(256, 256).astype(np.float32)
print(f"Shape: {data.shape}, Dtype: {data.dtype}")
```

## Step 2: Preview upload (validates before submitting)

`preview_upload()` checks naming convention, metadata schema, shape, and duplicates — without uploading anything.

```python
from quantem.data import preview_upload

errors = preview_upload(
    data,
    name="gold_nanoparticle",
    technique="hrtem",
    description="Gold nanoparticle HRTEM at 200 kV",
    contributor="Jane Doe",
)

if errors:
    print("Fix these errors before uploading:")
    for e in errors:
        print(f"  - {e}")
else:
    print("Validation passed — ready to upload!")
```

## What happens next?

1. Your data + metadata JSON are uploaded as a **Pull Request** on [HF Hub](https://huggingface.co/datasets/bobleesj/quantem-data/discussions)
2. A maintainer reviews the PR (naming, metadata, data quality)
3. Once merged, the dataset is available via `quantem.data.load("gold_nanoparticle")`

## Requirements

- Data must be shareable under an open license (CC-BY-4.0 recommended)
- Include instrument and sample information
- Follow the naming convention above
- All uploads go through PR review
