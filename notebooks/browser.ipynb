```python
from quantem.data import DataBrowser

browser = DataBrowser()
browser
```

## Using loaded data

After selecting a dataset and clicking **Load** in the widget above, access the data:

```python
# Check what's loaded
browser.summary()
```

```python
# Access the loaded array
if browser.data is not None:
    print(f"Shape: {browser.data.shape}")
    print(f"Dtype: {browser.data.dtype}")
    print(f"Name: {browser.loaded_name}")
    print(f"Metadata: {browser.metadata}")
```

## Filter by technique

Use the dropdown in the widget, or pass a technique at creation:

```python
# Show only 4D-STEM datasets
browser_4dstem = DataBrowser(technique="4dstem")
browser_4dstem
```

## Use with quantem.widget

Load a dataset and visualize it with quantem.widget:

```python
# Requires: pip install quantem-widget
if browser.data is not None:
    try:
        from quantem.widget import Show2D
        Show2D(browser.data, title=browser.loaded_name)
    except ImportError:
        print("Install quantem-widget to visualize: pip install quantem-widget")
```
