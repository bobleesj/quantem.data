## Using loaded data

After selecting a dataset and clicking **Load** in the widget above, access the data:

```python
browser.summary()
```

```python
if browser.data is not None:
    print(f"Shape: {browser.data.shape}")
    print(f"Dtype: {browser.data.dtype}")
    print(f"Name: {browser.loaded_name}")
    print(f"Metadata: {browser.metadata}")
```

## Use with quantem.widget

```python
# Visualize loaded data with quantem.widget (installed above on Colab)
if browser.data is not None:
    from quantem.widget import Show2D
    Show2D(browser.data, title=browser.loaded_name)
```
