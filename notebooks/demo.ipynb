```python
from quantem.data import available, info, load

# list all available datasets
available()
```

```python
# inspect metadata (no data download)
info("korean_sample_c1")
```

```python
from quantem.widget import Show2D

# load and display a 2D virtual image
Show2D(load("korean_sample_c1"), title="Korean Sample C1")
```

```python
# load with metadata
data, meta = load("korean_sample_c1", metadata=True)
print(f"Shape: {data.shape}, Technique: {meta['technique']}")
print(f"Description: {meta['description']}")
```

```python
# filter by technique
print("Images:", available(technique="image"))
```
