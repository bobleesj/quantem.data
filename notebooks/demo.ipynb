# quantem.data demo

Load real electron microscopy data from Hugging Face Hub and visualize with quantem.widget.

Works on Google Colab — no authentication needed.

```python
# install (run once)
# !pip install quantem-data quantem-widget
```

```python
from quantem.data import available, info, load

# list all available datasets
available()
```

```python
# inspect metadata (no data download)
info("korean_sample_c1")
```

```python
from quantem.widget import Show2D

# load and display a 2D virtual image
Show2D(load("korean_sample_c1"), title="Korean Sample C1")
```

```python
# load with metadata for calibrated display
data, meta = load("korean_sample_c1", metadata=True)
print(f"Shape: {data.shape}, Technique: {meta['technique']}")
print(f"Description: {meta['description']}")
```

```python
from quantem.widget import Show4DSTEM

# load 4D-STEM data (9 MB, cached after first download)
data = load("srtio3_lamella")
Show4DSTEM(data, scan_shape=(32, 32), title="SrTiO3 Lamella")
```

```python
# bright-field image from the same sample
Show2D(load("srtio3_bf"), title="SrTiO3 BF")
```

```python
# mean diffraction pattern
Show2D(load("srtio3_mean_dp"), title="SrTiO3 Mean DP", log_scale=True)
```

```python
# filter by technique
print("4D-STEM datasets:", available(technique="4dstem"))
print("Images:", available(technique="image"))
print("Diffraction:", available(technique="diffraction"))
```
